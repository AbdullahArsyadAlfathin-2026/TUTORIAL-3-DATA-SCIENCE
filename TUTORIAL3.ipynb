## 2.Evaluate if the data is good or not. Does it have enough samples?

```python
print(f"Total Rows: {len(df)}")
is_good = len(df) > 1000 and df['country'].nunique() > 10
print(f"Data Quality Assessment: {'Good dataset with high sample volume' if is_good else 'Insufficient sample size'}")
print(f"Countries Covered: {df['country'].nunique()}")
print(f"Years Covered: {df['year'].min()} to {df['year'].max()}\n")
```

## 3. how many features are there? is that enough?

```python
features_count = len(df.columns)
print(f"Number of Features (Columns): {features_count}")
print(f"Columns List: {df.columns.tolist()}")
print("Is it enough? Yes, 12 features covering demographic, economic, and temporal attributes are sufficient for EDA and regression modeling.\n")
```

## 4.how many observation are there? is that enough?

```python
samples_count = len(df.columns)
print(f"Number of Samples (Rows): {samples_count}")
print("Is it enough? Yes, 27,820 observations provide robust statistical significance.\n")
```

6 and 8. which columns and rows have missing (null) values

```python
missing = df.isnull().sum()
print("missing values in each column")
print(missing)
print("\nColumns with any missing values:")
print(missing[missing > 0])
print("\nNumber of columns with missing values" , (missing > 0).sum())
```

## 7. Which column is the feature, which is the target output?

```python
target = 'suicides/100k pop'
features = [c for c in df.columns if c != target]

print(f"Target Output (y): {target}")
print(f"Predictive Features (X): {features}\n")
```

## 9.What kind of question can be asked, should we ask. Does the dataset suits the question we are asking?

```python
import pandas as pd
from sklearn.datasets import load_iris

# Load data and define target
df = load_iris(as_frame=True).frame
target = "target"

# Evaluate dataset suitability for a classification question
print(f"Target Classes:\n{df[target].value_counts()}")
print(f"\nTotal Samples: {len(df)}")
print(f"\nData Types:\n{df.dtypes}")
print(
    "\n[Suitability Result]: The dataset is well-suited for a classification"
    " question due to sufficient sample size and balanced categorical"
    " targets."
)
```

```python
questions_eval = [
    {
        "question": "Which demographic group (age/gender) has the highest suicide risk?",
        "suitable": True,
        "reason": "Dataset contains explicit 'sex', 'age', and 'suicides/100k pop' columns."
    },
    {
        "question": "Does GDP per capita correlate with lower suicide rates across countries?",
        "suitable": True,
        "reason": "Dataset provides both 'gdp_per_capita ($)' and normalized rates."
    },
    {
        "question": "What is the impact of seasonal temperature changes on suicide rates?",
        "suitable": False,
        "reason": "Dataset does not contain monthly timestamps or climate/weather data."
    }
]

for q in questions_eval:
    status = "SUITABLE" if q["suitable"] else "UNSUITABLE"
    print(f"Question: {q['question']}")
    print(f"Status: {status}")
    print(f"Reason: {q['reason']}\n")
```
